# One rule

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report

class OneRClassifier:
    def __init__(self, n_bins=50):
        self.n_bins = n_bins
        self.best_feature = None
        self.best_rules = {}
        self.bins = None
        self.top_10_features_ = []  # (feature, accuracy) список

    def _bin_feature(self, values):
        """Квантильное разбиение признака на интервалы."""
        percentiles = np.linspace(0, 100, self.n_bins + 1)
        bins = np.percentile(values, percentiles)
        bins = np.unique(bins)
        binned = np.digitize(values, bins[1:-1], right=False)
        return binned, bins

    def fit(self, X, y):
        feature_scores = []

        for feature in X.columns:
            values = X[feature].values
            binned, bins = self._bin_feature(values)

            rules = {}
            for bin_idx in np.unique(binned):
                indices = np.where(binned == bin_idx)[0]
                if len(indices) == 0:
                    continue
                majority_class = pd.Series(y[indices]).mode()[0]
                rules[bin_idx] = majority_class

            preds = [rules.get(b, -1) for b in binned]
            acc = accuracy_score(y, preds)
            bal_acc = balanced_accuracy_score(y, preds)

            feature_scores.append((feature, acc, bal_acc, rules, bins))

        # сортировка по accuracy убыванию
        feature_scores.sort(key=lambda x: x[1], reverse=True)
        self.top_10_features_ = [(feat, round(acc, 4), round(bal_acc, 4)) for feat, acc, bal_acc, _, _ in feature_scores[:10]]

        # выбираем лучший
        best_feature, best_acc, best_bal_acc, best_rules, best_bins = feature_scores[0]
        self.best_feature = best_feature
        self.best_rules = best_rules
        self.bins = best_bins

    def predict(self, X):
        values = X[self.best_feature].values
        binned = np.digitize(values, self.bins[1:-1], right=False)
        return np.array([self.best_rules.get(b, -1) for b in binned])

    def report(self, X, y_true):
        y_pred = self.predict(X)
        acc = accuracy_score(y_true, y_pred)
        bal_acc = balanced_accuracy_score(y_true, y_pred)
        print(f"Лучшая переменная: {self.best_feature}")
        print(f"Правила (бин -> класс): {self.best_rules}")
        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}\n")
        print("Отчёт по классам:\n")
        print(classification_report(y_true, y_pred, zero_division=0))
        return acc, bal_acc

    def show_top_features(self):
        print("\nТоп-10 признаков по точности OneR:")
        for i, (feat, acc, bal_acc) in enumerate(self.top_10_features_, 1):
            print(f"{i}. {feat}: accuracy = {acc}, balanced_accuracy = {bal_acc}")
```

# Логистическая регрессия

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report

class LogisticRegressionSelector:
    def __init__(self):
        self.model = None
        self.top_10_features_ = []
        self.coefs_ = None

    def fit(self, X, y):
        self.model = LogisticRegression(
            max_iter=1000, multi_class='multinomial', solver='lbfgs', n_jobs=-1
        )
        self.model.fit(X, y)

        # Расчёт важности признаков по среднему модулю коэффициентов
        self.coefs_ = np.abs(self.model.coef_).mean(axis=0)
        feature_importance = pd.DataFrame({
            'Feature': X.columns,
            'Importance': self.coefs_
        })
        feature_importance.sort_values(by='Importance', ascending=False, inplace=True)
        self.top_10_features_ = list(zip(
            feature_importance['Feature'].head(10),
            np.round(feature_importance['Importance'].head(10), 4)
        ))

    def predict(self, X):
        return self.model.predict(X)

    def report(self, X, y_true, class_names=None):
        y_pred = self.predict(X)
        acc = accuracy_score(y_true, y_pred)
        bal_acc = balanced_accuracy_score(y_true, y_pred)
        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}\n")
        print("Отчёт по классам:\n")
        print(classification_report(y_true, y_pred, target_names=class_names, zero_division=0))
        return acc, bal_acc

    def show_top_features(self):
        print("\nТоп-10 признаков по важности (логистическая регрессия):")
        for i, (feat, imp) in enumerate(self.top_10_features_, 1):
            print(f"{i}. {feat}: importance = {imp}")
```

# k ближайших соседей

```python
import time
from tqdm import tqdm
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split

class KNNFullModelSelector:
    def __init__(self, n_neighbors=5):
        self.n_neighbors = n_neighbors
        self.model = None
        self.top_10_features_ = []
        self.base_acc = None
        self.base_bal_acc = None

    def fit(self, X, y):
        self.X = X
        self.y = y

        smote = SMOTE(random_state=42)
        X_resampled, y_resampled = smote.fit_resample(X, y)
        print("Обучение полноценной модели KNN...")
        self.model = KNeighborsClassifier(n_neighbors=self.n_neighbors)
        self.model.fit(X_resampled, y_resampled)
        print("Обучение завершено")

        # Базовая точность модели на полном наборе
        y_pred = self.model.predict(X)
        self.base_acc = accuracy_score(y, y_pred)
        self.base_bal_acc = balanced_accuracy_score(y, y_pred)

        # Оценка важности признаков: удаляем по одному и измеряем падение метрик
        importances = []
        cols = tqdm(X.columns, desc="Оценка признаков")
        for col in cols:
            X_reduced = X.drop(columns=[col])
            X_res_red, y_res_red = smote.fit_resample(X_reduced, y)
        
            model = KNeighborsClassifier(n_neighbors=self.n_neighbors)
            model.fit(X_res_red, y_res_red)
        
            y_pred_reduced = model.predict(X_reduced)
            acc = accuracy_score(y, y_pred_reduced)
            drop_in_acc = round(self.base_acc - acc, 4)
        
            importances.append((col, drop_in_acc))

        importances.sort(key=lambda x: x[1], reverse=True)
        self.top_10_features_ = importances[:10]

    def predict(self, X):
        return self.model.predict(X)

    def report(self, X_test, y_test):
        y_pred = self.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        bal_acc = balanced_accuracy_score(y_test, y_pred)
        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}\n")
        print("Отчёт по классам:\n")
        print(classification_report(y_test, y_pred, zero_division=0))
        return acc, bal_acc

    def show_top_features(self):
        print("\nТоп-10 признаков по важности (удаление признака ↓):")
        for i, (feat, delta) in enumerate(self.top_10_features_, 1):
            print(f"{i}. {feat}: снижение accuracy = {delta}")
```

# Наивный байесовский классификатор

```python
import numpy as np
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split

class NaiveBayesFullModelSelector:
    def __init__(self):
        self.model = None
        self.top_10_features_ = []
        self.base_acc = None
        self.base_bal_acc = None

    def fit(self, X, y):
        self.X = X
        self.y = y

        smote = SMOTE(random_state=42)
        X_resampled, y_resampled = smote.fit_resample(X, y)

        self.model = GaussianNB()
        self.model.fit(X_resampled, y_resampled)

        y_pred = self.model.predict(X)
        self.base_acc = accuracy_score(y, y_pred)
        self.base_bal_acc = balanced_accuracy_score(y, y_pred)

        importances = []
        for col in X.columns:
            X_reduced = X.drop(columns=[col])
            X_res_red, y_res_red = smote.fit_resample(X_reduced, y)
            model = GaussianNB()
            model.fit(X_res_red, y_res_red)
            y_pred_reduced = model.predict(X_reduced)
            acc = accuracy_score(y, y_pred_reduced)
            drop_in_acc = self.base_acc - acc
            importances.append((col, round(drop_in_acc, 4)))

        importances.sort(key=lambda x: x[1], reverse=True)
        self.top_10_features_ = importances[:10]

    def predict(self, X):
        return self.model.predict(X)

    def report(self, X_test, y_test):
        y_pred = self.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        bal_acc = balanced_accuracy_score(y_test, y_pred)
        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}\n")
        print("Отчёт по классам:\n")
        print(classification_report(y_test, y_pred, zero_division=0))
        return acc, bal_acc

    def show_top_features(self):
        print("\nТоп-10 признаков по важности (удаление признака ↓):")
        for i, (feat, delta) in enumerate(self.top_10_features_, 1):
            print(f"{i}. {feat}: снижение accuracy = {delta}")
```

# Деревья решений

```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split

class DecisionTreeFullModelSelector:
    def __init__(self, max_depth=None):
        self.max_depth = max_depth
        self.model = None
        self.top_10_features_ = []

    def fit(self, X, y):
        self.X = X
        self.y = y

        smote = SMOTE(random_state=42)
        X_resampled, y_resampled = smote.fit_resample(X, y)

        self.model = DecisionTreeClassifier(max_depth=self.max_depth, random_state=42)
        self.model.fit(X_resampled, y_resampled)

        importances = self.model.feature_importances_
        feature_importance = pd.DataFrame({
            'Feature': X.columns,
            'Importance': importances
        }).sort_values(by='Importance', ascending=False)

        self.top_10_features_ = feature_importance.head(10).values.tolist()

    def predict(self, X):
        return self.model.predict(X)

    def report(self, X_test, y_test):
        y_pred = self.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        bal_acc = balanced_accuracy_score(y_test, y_pred)
        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}\n")
        print("Отчёт по классам:\n")
        print(classification_report(y_test, y_pred, zero_division=0))
        return acc, bal_acc

    def show_top_features(self):
        print("\nТоп-10 признаков по важности (Decision Tree):")
        for i, (feat, imp) in enumerate(self.top_10_features_, 1):
            print(f"{i}. {feat}: importance = {imp:.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

class SVMFullModelSelector:
    def __init__(self, C=1.0):
        self.C = C
        self.model = None
        self.top_10_features_ = []
        self.scaler = StandardScaler()

    def fit(self, X, y):
        self.X = X
        self.y = y

        X_scaled = self.scaler.fit_transform(X)

        smote = SMOTE(random_state=42)
        X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

        self.model = LinearSVC(C=self.C, max_iter=5000, dual=False, random_state=42)
        self.model.fit(X_resampled, y_resampled)

        importances = np.abs(self.model.coef_).mean(axis=0)
        feature_importance = pd.DataFrame({
            'Feature': X.columns,
            'Importance': importances
        }).sort_values(by='Importance', ascending=False)

        self.top_10_features_ = feature_importance.head(10).values.tolist()

    def predict(self, X):
        X_scaled = self.scaler.transform(X)
        return self.model.predict(X_scaled)

    def report(self, X_test, y_test):
        y_pred = self.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        bal_acc = balanced_accuracy_score(y_test, y_pred)
        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}\n")
        print("Отчёт по классам:\n")
        print(classification_report(y_test, y_pred, zero_division=0))
        return acc, bal_acc

    def show_top_features(self):
        print("\n Топ-10 признаков по важности (SVM):")
        for i, (feat, imp) in enumerate(self.top_10_features_, 1):
            print(f"{i}. {feat}: importance = {imp:.4f}")
```

# Полносвязная нейронная сеть (multi layer perceptron)

```python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
import numpy as np
import pandas as pd

class MLPFullModelSelector:
    def __init__(self, input_dim, hidden_dim=64, epochs=20, batch_size=32, lr=0.001):
        self.input_dim = input_dim
        self.hidden_dim = hidden_dim
        self.epochs = epochs
        self.batch_size = batch_size
        self.lr = lr
        self.scaler = StandardScaler()
        self.model = None

    def build_model(self, num_classes):
        model = nn.Sequential(
            nn.Linear(self.input_dim, self.hidden_dim),
            nn.ReLU(),  
            nn.Dropout(0.3),
            nn.Linear(self.hidden_dim, self.hidden_dim // 2),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(self.hidden_dim // 2, num_classes) 
        )
        return model


    def fit(self, X, y):
        X_scaled = self.scaler.fit_transform(X)
        smote = SMOTE(random_state=42)
        X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

        X_tensor = torch.tensor(X_resampled, dtype=torch.float32)
        y_tensor = torch.tensor(y_resampled, dtype=torch.long)

        num_classes = len(np.unique(y))
        self.model = self.build_model(num_classes)
        criterion = nn.CrossEntropyLoss()
        optimizer = optim.Adam(self.model.parameters(), lr=self.lr)

        dataset = torch.utils.data.TensorDataset(X_tensor, y_tensor)
        loader = torch.utils.data.DataLoader(dataset, batch_size=self.batch_size, shuffle=True)

        self.model.train()
        for epoch in range(self.epochs):
            for batch_X, batch_y in loader:
                optimizer.zero_grad()
                outputs = self.model(batch_X)
                loss = criterion(outputs, batch_y)
                loss.backward()
                optimizer.step()

    def predict(self, X):
        X_scaled = self.scaler.transform(X)
        X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
        self.model.eval()
        with torch.no_grad():
            outputs = self.model(X_tensor)
            _, preds = torch.max(outputs, 1)
        return preds.numpy()

    def report(self, X_test, y_test):
        y_pred = self.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        bal_acc = balanced_accuracy_score(y_test, y_pred)
        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}\n")
        print("Отчёт по классам:\n")
        print(classification_report(y_test, y_pred, zero_division=0))
        return acc, bal_acc
```

# Ансамбль моделей

```python
import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report

class CustomEnsembleClassifier:
    def __init__(self, models, fit_models=True):
        self.models = models
        self.fitted_models = []
        self.fit_models = fit_models
        self.feature_names_in_ = None  # для отслеживания имён колонок

    def fit(self, X, y):
        if not self.fit_models:
            self.fitted_models = self.models
            print("Используем уже обученные модели, fit пропущен.")
            return self

        self.fitted_models = []
        self.feature_names_in_ = X.columns if hasattr(X, "columns") else None

        for name, model in self.models:
            print(f"Training {name}...")
            cloned_model = clone(model)
            cloned_model.fit(X, y)
            self.fitted_models.append((name, cloned_model))
        return self

    def predict(self, X):
        if not self.fitted_models:
            raise ValueError("Ensemble not fitted yet. Call fit() first.")

        predictions = np.array([model.predict(X) for _, model in self.fitted_models])
        majority_votes = np.apply_along_axis(lambda x: np.bincount(x).argmax(), axis=0, arr=predictions)
        return majority_votes

    def report(self, X, y_true, class_names=None, all_possible_classes=None):
        y_pred = self.predict(X)

        acc = accuracy_score(y_true, y_pred)
        bal_acc = balanced_accuracy_score(y_true, y_pred)

        print(f"Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}")

        used_classes = np.unique(y_true)

        if class_names is not None and all_possible_classes is not None:
            label_to_name = dict(zip(all_possible_classes, class_names))
            target_names = [str(label_to_name.get(c, str(c))) for c in used_classes]
        else:
            target_names = [str(c) for c in used_classes]

        print("Classification Report:\n")
        print(classification_report(
            y_true,
            y_pred,
            labels=used_classes,
            target_names=target_names,
            zero_division=0
        ))

        return acc, bal_acc

    def get_feature_importance(self, normalize=True):
        importances = []
        names = []

        for name, model in self.fitted_models:
            if hasattr(model, 'feature_importances_'):
                imp = model.feature_importances_
            elif hasattr(model, 'coef_'):
                imp = np.abs(model.coef_).mean(axis=0)
            else:
                print(f"[!] Модель '{name}' не поддерживает важность признаков.")
                continue

            importances.append(imp)
            names.append(name)

        if not importances:
            print("Нет моделей с поддержкой оценки важности.")
            return None

        # Среднее значение важностей по всем моделям
        avg_importance = np.mean(importances, axis=0)

        if self.feature_names_in_ is not None:
            importance_df = pd.DataFrame({
                'feature': self.feature_names_in_,
                'importance': avg_importance
            }).sort_values(by='importance', ascending=False)
        else:
            importance_df = pd.DataFrame({
                'feature': [f'feature_{i}' for i in range(len(avg_importance))],
                'importance': avg_importance
            }).sort_values(by='importance', ascending=False)

        if normalize:
            total = importance_df['importance'].sum()
            if total > 0:
                importance_df['importance'] /= total

        return importance_df
```
